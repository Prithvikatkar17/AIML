```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
insurance_data = pd.read_csv("insurance.csv")
```

```python
insurance_data
```

```python
 #visualization
plt.scatter(x=insurance_data["bmi"], y=insurance_data["charges"],c=insurance_data["smoker"].map({"yes": "red", "no": "blue"}))
```

```python
X=insurance_data.drop(columns=["region","charges"])
Y=insurance_data["charges"]

X["sex"]=X["sex"].map({"female":0,"male":1})
X["smoker"]=X["smoker"].map({"yes":1,"no":0})
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,random_state = 42)
```

```python
X_test.head()
```

```python
#traing model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
#predition
Y_pred = model.predict(X_test)
Y_pred
```

```python
Y_test
```

```python
#evaluatiion
from sklearn.metrics import r2_score


r2 = r2_score(Y_test,Y_pred)
print("r2_scored ",r2)

n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1 - ((1-r2) * (n-1) / (n-p-1))
print("adjusted r^2:", adjusted_r2)
```

```python
# one hot encoding
X=insurance_data.drop(columns=["charges"])
Y=insurance_data["charges"]

X = pd.get_dummies(X,columns =["region"],drop_first=False,dtype=int)

X["sex"]=X["sex"].map({"female":0,"male":1})
X["smoker"]=X["smoker"].map({"yes":1,"no":0})
```

```python
X.head()
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,random_state = 42)

model = LinearRegression()
model.fit(X_train,Y_train)
Y_pred = model.predict(X_test)
```

```python
r2 = r2_score(Y_test , Y_pred)
print("r2_score = ", r2)
```

```python
# interaction feature
X=insurance_data.drop(columns=["charges"])
Y=insurance_data["charges"]

X = pd.get_dummies(X,columns =["region"],drop_first=True,dtype=int)



X["sex"]=X["sex"].map({"female":0,"male":1})
X["smoker"]=X["smoker"].map({"yes":1,"no":0})

X["age_smoker"]= X["age"] * X["smoker"]
X["bmi_smoker"] = X["bmi"] * X["smoker"]

X.head()
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.2,random_state = 42)

model = LinearRegression()
model.fit(X_train,Y_train)
Y_pred = model.predict(X_test)
r2 = r2_score(Y_test , Y_pred)
print("r2_score = ", r2)
```
