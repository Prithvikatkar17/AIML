```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df2 = df.copy()
```

```python
# Renaming
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income"]
df2
```

```python

df2.rename(index={0: "first"})
```

```python
df2.rename(columns={"Income": "Salary"})
```

```python
df2.sort_values("Salary")
```

```python
# Sorting - values & index
df2.sort_values("Salary")                     # sort values in ascending
df2.sort_values("Salary", ascending=False)    # sort values in descending
df2.sort_values(["Age", "Salary"])            # sorts for age, if age is same then sorts for income 

sorted_df2 = df2.sort_values(["Age", "Salary"])
sorted_df2.sort_index()
```

```python
# Reset Index
sorted_df2.reset_index()
sorted_df2.reset_index(drop=True)     # to drop original index vals
```

```python
# Ranking
df2["Ranking"] = df2["Salary"].rank(ascending=False, method="dense")
```
