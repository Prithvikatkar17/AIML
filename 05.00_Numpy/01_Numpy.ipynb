```python
import numpy as np
import time
```

```python
size = 10_000_000
py_list = list(range(size))
start = time.time()
sq_list = [x**2  for x in py_list]
end = time.time()
print(f"time for python list = {end - start} seconds")

np_arr = np.array(py_list)
start = time.time()
sq_arr = np_arr**2
end = time.time()
print(f"time for np array = {end - start} seconds")
```

```python
arr = np.array([1,2,3,4,5])
print(arr ,type(arr) , arr.shape)

arr2 = np.array([1,2,3,4,5,"np"])
print(arr2 ,type(arr2) , arr2.shape)

```

```python
arr1 = np.zeros((3,3),dtype="int64")
print(arr1 , arr1.shape)

arr2 = np.ones((3,3),dtype="int64")
print(arr2 , arr2.shape)

arr3 = np.full((3,3),100)
print(arr3 , arr3.shape)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr , arr.shape)
```

```python
reshapedArr = arr.reshape((3,2))
print(reshapedArr,reshapedArr.shape)

flattened = arr.flatten()
print(flattened ,flattened.shape)
```

```python
arr = np.array([1,2,3,4,5,6,7])

idx = [2,5]
print(arr[idx])
print(arr[arr >2])
print(arr[arr % 2 == 0])
print(arr[arr % 2 != 0])
```
